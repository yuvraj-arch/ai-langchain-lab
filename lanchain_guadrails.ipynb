```python
from typing import Any

from langchain.agents.middleware import AgentMiddleware, AgentState, hook_config

from langgraph.runtime import Runtime
```

```python
class ContentFilterMiddleware(AgentMiddleware):
    def __init__(self, agent: Agent, content_filter: ContentFilter):
        super().__init__(agent)
        self.content_filter = content_filter

    def _call(self, agent_state: AgentState, messages: List[BaseMessage]) -> List[BaseMessage]:
        filtered_messages = self.content_filter.filter_messages(messages)
        return filtered_messages

    def _plan(self, agent_state: AgentState, messages: List[BaseMessage]) -> List[BaseMessage]:
        filtered_messages = self.content_filter.filter_messages(messages)
        return filtered_messages
    
    def _execute(self, agent_state: AgentState, messages: List[BaseMessage]) -> List[BaseMessage]:
        filtered_messages = self.content_filter.filter_messages(messages)
        return filtered_messages

    def _finish(self, agent_state: AgentState, messages: List[BaseMessage]) -> List[BaseMessage]:
        filtered_messages = self.content_filter.filter_messages(messages)
        return filtered_messages
    
    def _stream(self, agent_state: AgentState, messages: List[BaseMessage]) -> List[BaseMessage]:
        filtered_messages = self.content_filter.filter_messages(messages)
        return filtered_messages
    
    def _stream_finish(self, agent_state: AgentState, messages: List[BaseMessage]) -> List[BaseMessage]:
        filtered_messages = self.content_filter.filter_messages(messages)
        return filtered_messages
```

```python
def search_tool(query: str) -> str:
    """
    Search the web for information.
    """
    return f"Searching the web for: {query}"
```

```python
# Use the custom guardrail
from langchain.agents import create_agent

agent = create_agent(
    model="claude-sonnet-4-5",
    tools=[search_tool],
    middleware=[
        ContentFilterMiddleware(
            banned_keywords=["hack", "exploit", "malware"]
        ),
    ],
)

# This request will be blocked before any processing
result = agent.invoke({
    "messages": [{"role": "user", "content": "How do I hack into a database?"}]
})
```
